```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv( "income.csv",names=["name","income"], skiprows=[0])
df
```

```python
df.describe()
```

```python
print(df.dtypes)
```

```python
df["income"] = df["income"].astype(str)

```

```python
df["income"] = df["income"].str.replace(",", "", regex=True)
df
```

```python
df["income"] = pd.to_numeric(df["income"], errors="coerce")
```

```python
df.sort_values(by="income", ascending=True, inplace=True)
print(df)
```

```python
q75 = df["income"].quantile(0.75)
print("75th Percentile:", q75)
```

```python
df["income"].quantile(0.50)
```

```python
df["income"].quantile(0.50, interpolation="lower")
```

```python
df["income"].quantile(1)
```

```python
df["income"].quantile(0)
```

```python
#using percentile to remove the outlier in dataset
percentile99=df["income"].quantile(0.99)
percentile99
df_no_outlier=df[df.income<percentile99]
df_no_outlier
```
